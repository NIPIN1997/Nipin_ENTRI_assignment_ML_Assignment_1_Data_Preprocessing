### Finding outliers

### Checking for missing values

```python
data_copy.isna().sum()
```

**Missing values have been identified in the following columns.**
1. company_name (categorical column)
2. employee_age (numerical column)
3. salary (numerical column)
4. place (categorical column)

***Replacing the missing values in the categorical columns with the mode value of the respective columns and replacing the missing values in numerical columns with either the mean or median values of the respective columns.***


#### Replacing the missing values in the company_name column with the mode value of the column

```python
mode_imputer=SimpleImputer(strategy='most_frequent')
data_copy['company_name']=mode_imputer.fit_transform(data_copy[['company_name']]).ravel()
```

#### Replacing the missing values in the place column with the mode value of the column

```python
data_copy['place']=mode_imputer.fit_transform(data_copy[['place']]).ravel()
```

#### Checking for outliers in the employee_age column

```python
# plotting a box plot
fig, ax=plt.subplots(figsize=(10,10))
sns.boxplot(
    data=data_copy,
    y='employee_age',
    color='yellow'
)
ax.set_title("Box plot for employee_age column")
plt.show()
```

***Since no outliers have been identified in the employee_age column, the missing values in the employee_age column are filled with the mean value of the column.***

#### Replacing the missing values in the employee_age column with the mean value of the column

```python
mean_imputer=SimpleImputer(strategy='mean')
data_copy['employee_age']=mean_imputer.fit_transform(data_copy[['employee_age']])
```

#### Checking for outliers in the salary column

```python
# plotting a box plot
fig, ax=plt.subplots(figsize=(10,10))
sns.boxplot(
    data=data_copy,
    y='salary',
    color='yellow'
)
ax.set_title("Box plot for salary column")
plt.show()
```

***Since no outliers have been identified in the salary column, the missing values in the salary column are filled with the mean value of the column.***

#### Replacing the missing values in the salary column with the mean value of the column

```python
data_copy['salary']=mean_imputer.fit_transform(data_copy[['salary']])
```

#### Checking if there are any more missing values

```python
data_copy.isna().sum()
```

***No more missing values are there in the dataset.***

## Data Encoding

```python
data_copy.info()
```

***The categorical columns are company_name, place, country.***

#### Identifying unique values in the company_name column

```python
data_copy['company_name'].unique()
```

***company_name column has 4 unique values, so using label encoding instead of one-hot encoding as one-hot encoding will result in an increase in the number of features.***

```python
label_encoder=LabelEncoder()
data_copy['company_name']=label_encoder.fit_transform(data_copy['company_name'])
```

```python
data_copy.head(10)
```

#### Identifying unique values in the place column

```python
data_copy['place'].unique().tolist()
```

***place column has 11 unique values, so using label encoding instead of one-hot encoding as one-hot encoding will result in an increase in the number of features.***

```python
data_copy['place']=label_encoder.fit_transform(data_copy['place'])
```

```python
data_copy.head(10)
```

#### Identifying unique values in the country column

```python
data_copy['country'].unique()
```

***country column has only 1 unique value, so using one-hot encoding.***

```python
data_copy=pd.get_dummies(data=data_copy,columns=['country'])
```

```python
data_copy.head(10)
```
